```python
import pandas as pd

# 1. Upload the file in Colab first (Files sidebar → Upload),
#    or if it's already there, just load it:
df = pd.read_csv('fbref ball carrying - all.csv')

# 2. Drop all repeated "Rk" rows except the first one
mask = df.iloc[:, 0] == 'Rk'
if mask.any():
    first_idx = mask.idxmax()
    mask.loc[first_idx] = False          # keep the first Rk row

df = df[~mask].copy()

# 3. Promote that first row to real column headers
df.columns = df.iloc[0]
df = df[1:].reset_index(drop=True)

# 4. Done — check the result
print(df.shape)
df.head()
```

```python
# After you've cleaned the Rk rows and set headers (from earlier code)

# Convert key columns to numbers
for col in ['90s', 'Touches', 'Carries', 'TotDist', 'PrgDist', 'PrgC', '1/3', 'CPA']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Filter: keep players with meaningful progressive carrying
# - At least 5 full matches (90s >= 5)
# - At least 10 progressive carries OR 800+ progressive distance
df_filtered = df[
    (df['90s'] >= 5) &
    ((df['PrgC'] >= 10) | (df['PrgDist'] >= 800))
].copy()

print(f"Players before: {len(df)}")
print(f"Players after:  {len(df_filtered)}")
print(f"Removed:        {len(df) - len(df_filtered)}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Make charts look better
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(
    data=df_filtered,
    x='PrgC',
    y='PrgDist',
    hue='Pos',
    size='90s',          # bubble size = minutes played
    sizes=(20, 200),
    alpha=0.7
)
plt.title('Progressive Carries vs Progressive Distance')
plt.xlabel('Progressive Carries (PrgC)')
plt.ylabel('Progressive Distance (PrgDist)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Force both axes to be numeric
df_filtered['PrgC'] = pd.to_numeric(df_filtered['PrgC'], errors='coerce')
df_filtered['Succ'] = pd.to_numeric(df_filtered['Succ'], errors='coerce')

# Drop any rows that failed to convert
df_plot = df_filtered.dropna(subset=['PrgC', 'Succ'])

# Recreate the chart with explicit numeric axes
plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df_plot,
    x='PrgC',
    y='Succ',
    hue='Pos',
    size='90s',
    sizes=(30, 250),
    alpha=0.7
)

plt.title('Progressive Carries vs Successful Dribbles')
plt.xlabel('Progressive Carries (PrgC)')
plt.ylabel('Successful Dribbles (Succ)')

# Force clean numeric ticks
plt.yticks(range(0, int(df_plot['Succ'].max()) + 10, 5))
plt.xticks(range(0, int(df_plot['PrgC'].max()) + 20, 10))

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x='Weighted',
    y='Context',
    s=80,
    alpha=0.7
)

# Only label standout players
for _, row in df.iterrows():
    if row['Weighted'] > 9.5 or row['Context'] > 0.95:
        plt.annotate(
            row['Player'],
            (row['Weighted'], row['Context']),
            fontsize=8,
            alpha=0.85,
            xytext=(4, 4),
            textcoords='offset points'
        )

plt.title('Weighted vs Context')
plt.xlabel('Weighted')
plt.ylabel('Context')
plt.tight_layout()
plt.show()
```
